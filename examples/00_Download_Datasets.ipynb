# Download Datasets
Run this notebook first to fetch the real-world datasets that will be used by the other advanced notebooks. The datasets are saved into the `examples/datasets/` folder.


```python
import os
import pandas as pd
from sklearn.datasets import fetch_california_housing, load_diabetes
import yfinance as yf

# Create datasets directory
os.makedirs('datasets', exist_ok=True)
```

### 1. California Housing Dataset (Regression/Stats)


```python
print('Downloading California Housing...')
cal_housing = fetch_california_housing(as_frame=True)
df_cal = cal_housing.frame
df_cal.to_csv('datasets/california_housing.csv', index=False)
print(f'Saved california_housing.csv ({len(df_cal)} rows)')
```

### 2. Diabetes Dataset (EDA/Outliers)


```python
print('Downloading Diabetes Dataset...')
diabetes = load_diabetes(as_frame=True)
df_diab = diabetes.frame
df_diab.to_csv('datasets/diabetes.csv', index=False)
print(f'Saved diabetes.csv ({len(df_diab)} rows)')
```

### 3. Apple Stock Data (Time Series)


```python
print('Downloading AAPL Stock Data...')
ticker = yf.Ticker('AAPL')
hist = ticker.history(start='2022-01-01', end='2023-12-31')
hist = hist.reset_index()
# Strip timezone formatting for compatibility
hist['Date'] = hist['Date'].dt.tz_localize(None)
hist.to_csv('datasets/aapl_stock.csv', index=False)
print(f'Saved aapl_stock.csv ({len(hist)} rows)')
```
