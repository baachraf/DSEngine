# Zero-Code Remote Database Loading

DSEngine automatically supports SQL database loading out of the box. By supplying a valid SQL connection string as `data.source`, DSEngine bypasses local file lookup and authenticates with the remote (or local) database directly.

In this notebook, we use the declarative `configs/09_database_loading.yml` to automatically connect to `sqlite:///../examples/datasets/demo.db` and query the `employees` table without writing any python connectors.

## 1 — The YAML Configuration

Take a look at how minimal the configuration is. The `loader_params` simply specify which table configuration to load, or you can even provide a raw `query: "SELECT * FROM employees WHERE age > 30"`.

```yaml
db_experiment:
  data:
    source: "sqlite:///../examples/datasets/demo.db"
    loader_params:
      table: "employees"
  
  output:
    path: "outputs/db_loading"
    format: ["html", "json"]

  steps:
    - name: "summary_stats"
      type: "summary"
```

````python
from ds_engine.utils import pipeline_runner

# Run the database extraction and analysis purely via the YAML configuration
pipeline_runner.run('../configs/09_database_loading.yml', 'db_experiment')
````
