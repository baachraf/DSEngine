# Basic Exploratory Data Analysis (Declarative)
This notebook acts as an interactive runner for DSEngine. It strictly follows the declarative philosophy: it uses the pre-existing YAML configuration defined in `configs/01_basic_eda.yml`. 


```python
# Load and display the experiment configuration
with open('../configs/01_basic_eda.yml', 'r') as f:
    print(f.read())
```

### Execute the Pipeline

```python
import sys, os, glob
from IPython.display import HTML, display

sys.path.insert(0, os.path.abspath('..'))
from ds_engine.utils import pipeline_runner

# 1. Run the pipeline based purely on the pre-existing YAML config
config_path = '../configs/01_basic_eda.yml'
experiment_name = '01_Basic_EDA_Declarative'

print(f"Executing pipeline for '{experiment_name}' using: {config_path}...")
pipeline_runner.run(config_path, experiment_name)
```

### View the Report

```python
# 2. Find and display the generated HTML report
output_dirs = sorted(glob.glob(f'./outputs/{experiment_name}/*'))

if output_dirs:
    latest_report = os.path.join(output_dirs[-1], 'report.html')
    if os.path.exists(latest_report):
        display(HTML(latest_report))
    else:
        print("Report not found.")
else:
    print("No output directory found.")
```
